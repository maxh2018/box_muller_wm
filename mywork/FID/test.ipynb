```python
data_coco_path = "/data/FinAi_Mapping_Knowledge/TOG3/maxiaohui/cache/FID数据集/val2017"
ann_coco_path = "/data/FinAi_Mapping_Knowledge/TOG3/maxiaohui/cache/FID数据集/annotations/captions_val2017.json"
data_Flickr8k_path = "/data/FinAi_Mapping_Knowledge/TOG3/maxiaohui/cache/FID数据集/Flicker8k_Dataset"
ann_Flickr8k_path = "/data/FinAi_Mapping_Knowledge/TOG3/maxiaohui/cache/FID数据集/Flickr8k.token.txt"
```

```python
import json
import os
```

```python
#列出路径下所有图片文件的名称

def listdir(path):
    list_name = []
    for file in os.listdir(path):
        file_path = os.path.join(path, file)
        if os.path.isdir(file_path):
            listdir(file_path)
        else:
            list_name.append(file)
    return list_name
coco_files = listdir(data_coco_path)
```

```python
print(coco_files[:10])
```

```python
with open(ann_coco_path, 'r', encoding='utf-8') as f:
    ann_coco = json.load(f)
    print(ann_coco.keys())
    print(ann_coco['annotations'][0:10])
coco_anns = ann_coco['annotations']
```

```python
img_id_to_captions = {}
for ann in coco_anns:
    img_id = ann['image_id']
    caption = ann['caption']
    if img_id not in img_id_to_captions:
        img_id_to_captions[img_id] = []
    img_id_to_captions[img_id].append(caption)
```

```python
print(len(coco_files))
print(len(img_id_to_captions.keys()))
print(len(coco_anns))
print(img_id_to_captions.keys())
```

```python
partial_captions = []
for file_name in coco_files:
    img_id = int(file_name.split('.')[0][-6:])
    
    if img_id in img_id_to_captions:
        captions = img_id_to_captions[img_id]
        #选择长度最长的caption
        caption = sorted(captions, key=lambda x: len(x), reverse=True)[0]
        partial_captions.append(caption)
    else:
        print(img_id)
        raise ValueError('img_id not found in annotations')
```

```python
print(partial_captions[:10])
print(len(partial_captions))
```

```python
#保存到文件
with open('coco_partial_captions_as_prompts.json', 'w', encoding='utf-8') as f:
    for caption in partial_captions:
        d = {"Prompt": caption}
        f.write(json.dumps(d, ensure_ascii=False)+'\n')
```

```python
with open(ann_Flickr8k_path, 'r', encoding='utf-8') as f:
    ann_Flickr8k = f.readlines()
    ann_Flickr8k = [x.strip() for x in ann_Flickr8k]
    print(ann_Flickr8k[:10])
```

```python
img_name_to_captions = {}
for ann in ann_Flickr8k:
    img_name, caption = ann.strip().split('\t')
    img_key = img_name.split('.')[0]
    if img_key not in img_name_to_captions:
        img_name_to_captions[img_key] = []
    img_name_to_captions[img_key].append(caption)
    
print(len(img_name_to_captions.keys()))
print(len(ann_Flickr8k))
```

```python
Flickr8k_files = listdir(data_Flickr8k_path)
print(Flickr8k_files[:10])
print(len(Flickr8k_files))
```

```python
partial_captions = []
for file_name in Flickr8k_files:
    img_key = file_name.split('.')[0]
    if img_key in img_name_to_captions:
        captions = img_name_to_captions[img_key]
        caption = sorted(captions, key=lambda x: len(x), reverse=True)[0]
        partial_captions.append(caption)
    else:
        print(img_key)
        raise ValueError('img_key not found in annotations')
    
print(partial_captions[:10])
print(len(partial_captions))
```

```python
with open('Flickr8k_partial_captions_as_prompts.json', 'w', encoding='utf-8') as f:
    for caption in partial_captions:
        d = {"Prompt": caption}
        f.write(json.dumps(d, ensure_ascii=False)+'\n')
```
